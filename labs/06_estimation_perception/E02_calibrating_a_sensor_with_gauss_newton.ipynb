## Map: nonlinear least squares in five lines

Find parameters $x$ that minimize $\sum_i r_i(x)^2$. Linearize around the current guess with the Jacobian
$J_{ij} = \partial r_i / \partial x_j$ and solve the normal equations for the step:
$$J^\top J\,\Delta = -J^\top r, \qquad x \leftarrow x + \Delta$$
Repeat until the step is tiny. (Levenberg-Marquardt adds $\lambda I$ to $J^\top J$ for robustness far from the answer.)

**Accelerometer calibration:** hold the board still in many orientations. At rest the true acceleration has magnitude
$g$, so with per-axis scale $s$ and bias $b$:
$$a_{true} = \mathrm{diag}(s)\,(a_{raw} - b), \qquad r_i = \lVert \mathrm{diag}(s)(a_{raw,i} - b) \rVert - g$$
Six unknowns ($s_x, s_y, s_z, b_x, b_y, b_z$) need at least six well-spread orientations; use more to average noise.

A numerical Jacobian by forward differences is fine here: $J_{:,j} \approx \dfrac{r(x + h e_j) - r(x)}{h}$.
